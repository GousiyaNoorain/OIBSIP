```python
Q1 = df[['Age', 'entryEXAM', 'studyHOURS', 'Python', 'DB']].quantile(0.25)
Q3 = df[['Age', 'entryEXAM', 'studyHOURS', 'Python', 'DB']].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower bounds:")
print(lower_bound)

print("\nUpper bounds:")
print(upper_bound)
```

```python
numeric_cols = ['Age', 'entryEXAM', 'studyHOURS', 'Python', 'DB']

outlier_mask = (
    (df[numeric_cols] < lower_bound) |
    (df[numeric_cols] > upper_bound)
).any(axis=1)

df[outlier_mask]
```

```python
df[df['Python'].isnull()]
```

```python
df['Python'] = df['Python'].fillna(df['Python'].median())
```

```python
df['Python'].isnull().sum()
```

```python
df['gender'] = df['gender'].replace({
    'F': 'Female',
    'female': 'Female',
    'M': 'Male',
    'male': 'Male'
})
```

```python
df['gender'].value_counts()
```

```python
df['country']=df['country'].replace({
    'norway':'Norway',
    'Norge':'Norway',
    'Rsa':'South Africa',
    'UK':'United Kingdo'
})
```

```python
df['country'].value_counts()
```

```python
df[df['country']=='Somali']
```

```python
df['country']=df['country'].replace({
    'Somali':'Somalia'
})
```

```python
df['country'].value_counts()
```

```python
df['residence']=df['residence'].replace({
    'BI-Residence':'BI Residence',
    'BIResidence':'BI Residence',
    'BI_Residence':'BI Residence'
})
```

```python
df['residence'].value_counts()
```

```python
df['prevEducation'] = df['prevEducation'].replace({
    'HighSchool': 'High School',
    'Barrrchelors': 'Bachelors',
    'diploma': 'Diploma',
    'DIPLOMA': 'Diploma',
    'Diplomaaa': 'Diploma'
})
```

```python
df['prevEducation'].value_counts()
```

```python
df['Python']=df['Python'].astype(int)
```

## Data Cleaning Methodology

The dataset was inspected and cleaned to improve data quality, consistency, and reliability.

### Missing Values
- 2 missing values were found in the `Python` column.
- Missing values were replaced using the median Python score (81).

### Duplicate Records
- No duplicate rows were identified.
- Therefore, no rows were removed due to duplication.

### Inconsistent Values
Inconsistent categorical values were standardized, including:
- Gender variations → `Female` / `Male`
- Country variations → standardized country names
- Residence variations → `BI Residence`
- Education variations → standardized education categories

### Outlier Detection
The IQR method was used to identify potential numerical outliers. The flagged observations were retained because their values were plausible and there was no clear evidence of data-entry errors.

### Data Types
Numerical columns were verified and corrected where necessary. The `Python` column was converted from `float64` to `int64` after missing values were handled.
